```python
import random
```

```python
def create_board():
    return [[" " for _ in range(3)] for _ in range(3)]
```

```python
def print_board(board):
    for r in range(3):
        print("  | ".join(board[r]))
        if r < 2:
            print("---+----+---")
```

```python
def available_moves(board):
    moves = []
    for r in range(3):
        for c in range(3):
            if board[r][c] == " ":
                moves.append((r, c))
    return moves
```

```python
def make_move(board, move, player):
    r, c = move
    if board[r][c] != " ":
        return False
    board[r][c] = player
    return True
```

```python
def check_winner(board):
    lines = []

    # rader
    for r in range(3):
        lines.append(board[r])

    # kolumner
    for c in range(3):
        lines.append([board[0][c], board[1][c], board[2][c]])

    # diagonaler
    lines.append([board[0][0], board[1][1], board[2][2]])
    lines.append([board[0][2], board[1][1], board[2][0]])

    for line in lines:
        if line[0] != " " and line[0] == line[1] == line[2]:
            return line[0]

    if len(available_moves(board)) == 0:
        return "draw"

    return None
```

```python
def human_player(board, player):
    while True:
        try:
            text = input(f"Spelare {player}, välj rad kolumn (0–2), ex '1 2': ")
            r, c = map(int, text.split())

            if not (0 <= r <= 2 and 0 <= c <= 2):
                print("❌ Måste vara mellan 0 och 2")
                continue

            if board[r][c] != " ":
                print("❌ Rutan är upptagen")
                continue

            return (r, c)

        except:
            print("❌ Fel input, försök igen")
```

```python
def random_player(board, player):
    return random.choice(available_moves(board))
```

```python
def play_game(player_O_func, player_X_func, verbose=True):
    board = create_board()
    current = "O"

    if verbose:
        print("Start:")
        print_board(board)
        print()

    while True:
        move = player_O_func(board, "O") if current == "O" else player_X_func(board, "X")

        ok = make_move(board, move, current)
        if not ok:
            winner = "X" if current == "O" else "O"
            print(f"{current} gjorde ogiltigt drag → {winner} vinner")
            return winner

        if verbose:
            print(f"{current} spelar {move}")
            print_board(board)
            print()

        result = check_winner(board)
        if result is not None:
            print("Resultat:", result)
            return result

        current = "X" if current == "O" else "O"
```

```python
play_game(human_player, random_player, verbose=True)
```
