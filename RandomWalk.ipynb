```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
def random_walk(probabilities):
    """
    Perform a 2D random walk with directional probabilities.
    
    Parameters:
    -----------
    probabilities : list of 4 floats
        Probabilities for [up, right, down, left] movements.
        Should sum to 1.0
    
    Returns:
    --------
    xLocations : list
        List of x coordinates visited
    yLocations : list
        List of y coordinates visited
    """
    # Movement offsets: up, right, down, left
    xm = [0, 1, 0, -1]
    ym = [1, 0, -1, 0]
    
    # Starting position
    x, y = 0, 0
    xLocations = [x]
    yLocations = [y]
    actions=[]
    # Random walk until boundary is hit
    while x <= 200 and y <= 200:
        # Pick direction using numpy.random.choice
        direction = np.random.choice(4, p=probabilities)
        actions.append(direction)        
        x += xm[direction]
        y += ym[direction]
        
        xLocations.append(x)
        yLocations.append(y)
    
    return xLocations, yLocations,actions

xLoc,yLoc,actions=random_walk([0.25, 0.25, 0.25, 0.25])
plt.plot(xLoc,yLoc,'k')
plt.show()
```

```python
# Movement offsets: up, right, down, left
xLoc,yLoc,actions=random_walk([0.0, 1.0, 0.0, 0.0])
plt.plot(xLoc,yLoc,'k')
plt.show()
```

```python
# Movement offsets: up, right, down, left
for i in range(4):
    print(i,actions.count(i))
```

```python
policy=np.array([0.25, 0.25, 0.25, 0.25])
for i in range(100):
    xLoc,yLoc,actions=random_walk(policy)
    dummy=np.bincount(actions, minlength=4)
    policy=dummy/sum(dummy)
    print(i,len(actions))
```
